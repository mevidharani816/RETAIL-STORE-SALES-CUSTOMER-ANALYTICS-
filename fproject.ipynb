```python
tsd=fdata1["order_date"] = pd.to_datetime(fdata1["order_date"])
tsd
```

```python
rfm = fdata.merge(fdata1, on="order_id")
rfm
```

```python
snapshot_date = fdata1["order_date"].max() + pd.Timedelta(days=1)

rfm_data = rfm.groupby("customer_id").agg({
    "order_date": lambda x: (snapshot_date - x.max()).days,
    "order_id": "count",
    "total_amount": "sum"
})

rfm_data.columns = ["Recency", "Frequency", "Monetary"]

rfm_data.head()
```
